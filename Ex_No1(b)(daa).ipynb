```python
import time
import random
import sys

def interpolation_search(arr, target):
    """
    Interpolation Search Algorithm
    Time Complexity: O(log log n) average, O(n) worst case
    Space Complexity: O(1)
    """
    low, high = 0, len(arr) - 1
    comparisons = 0

    while low <= high and arr[low] <= target <= arr[high]:
        comparisons += 1
        if low == high:
            if arr[low] == target:
                return low, comparisons
            return -1, comparisons

        # Interpolation formula
        if arr[high] == arr[low]: # Avoid division by zero if all elements are same
            if arr[low] == target:
                return low, comparisons
            else:
                return -1, comparisons
        pos = low + int(((target - arr[low]) * (high - low))
                        / (arr[high] - arr[low]))

        # Ensure pos is within bounds
        pos = max(low, min(high, pos))

        if arr[pos] == target:
            return pos, comparisons
        elif arr[pos] < target:
            low = pos + 1
        else:
            high = pos - 1

    return -1, comparisons

def binary_search(arr, target):
    """Binary Search for comparison"""
    low, high = 0, len(arr) - 1
    comparisons = 0
    while low <= high:
        comparisons += 1
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid, comparisons
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

def performance_analysis():
    sizes = [1000, 5000, 10000, 50000, 100000]
    print(f"{'Size':>10} {'IS Time(ms)':>14} {'BS Time(ms)':>14} "
          f"{'IS Comparisons':>16} {'BS Comparisons':>16}")
    print('-' * 75)

    for size in sizes:
        arr = sorted(random.sample(range(size * 10), size))

        # Ex. No. 1                   Implementation and Performance Analysis of Interpolation Search              CS5303 – DAA Lab
        # Chennai Institute of Technology  |  Dept. of CSE  |  Page 5
        target = arr[random.randint(0, size - 1)]

        # Interpolation Search timing
        start = time.perf_counter()
        for _ in range(100):
            idx_is, comp_is = interpolation_search(arr, target)
        is_time = (time.perf_counter() - start) / 100 * 1000

        # Binary Search timing
        start = time.perf_counter()
        for _ in range(100):
            idx_bs, comp_bs = binary_search(arr, target)
        bs_time = (time.perf_counter() - start) / 100 * 1000

        print(f"{size:>10} {is_time:>14.4f} {bs_time:>14.4f} "
              f"{comp_is:>16} {comp_bs:>16}")

# --- Main ---
arr = [2, 5, 10, 15, 23, 35, 48, 60, 75, 90, 105, 120]
target = 35
idx, comps = interpolation_search(arr, target)
print(f"Array: {arr}")
print(f"Searching for: {target}")
print(f"Found at index: {idx}, Comparisons: {comps}")
print()
performance_analysis()

# --- Student Roll Number Analysis ---
print('\n--- Student Roll Number Search Analysis ---')
roll_numbers_array = list(range(1, 10001))
# Select a random target roll number
target_roll_number = random.randint(1, 10000)

print(f"Array size: {len(roll_numbers_array)}")
print(f"Searching for roll number: {target_roll_number}")

# Interpolation Search for roll numbers
is_idx, is_comps = interpolation_search(roll_numbers_array, target_roll_number)
print(f"Interpolation Search: Found at index {is_idx}, Comparisons: {is_comps}")

# Binary Search for roll numbers
bs_idx, bs_comps = binary_search(roll_numbers_array, target_roll_number)
print(f"Binary Search:      Found at index {bs_idx}, Comparisons: {bs_comps}")

print(f"Comparison:")
print(f"  Interpolation Search required {is_comps} probes.")
print(f"  Binary Search required {bs_comps} probes.")
```
