## QuickSelect Algorithm

QuickSelect is a selection algorithm to find the k-th smallest element in an unsorted list. It is a variant of the QuickSort algorithm. Instead of recurring on both sides, it only recurs on one side based on the position of the pivot.

```python
import random

# Test QuickSelect with a random array of 1000 integers
arr_size = 1000
random_arr = [random.randint(1, 10000) for _ in range(arr_size)]

# To ensure QuickSelect doesn't modify the original list for subsequent calls, create a copy

# Find k=1 (minimum element)
arr_copy_min = list(random_arr)
min_element = quick_select(arr_copy_min, 0, arr_size - 1, 1)
print(f"Original Array size: {arr_size}")
print(f"Minimum element (k=1): {min_element} (Verified: {min(random_arr)})")

# Find k=n (maximum element)
arr_copy_max = list(random_arr)
max_element = quick_select(arr_copy_max, 0, arr_size - 1, arr_size)
print(f"Maximum element (k=n): {max_element} (Verified: {max(random_arr)})")

# Find k=n//2 (median element)
# For even size arrays, median is often defined as the average of the two middle elements.
# QuickSelect finds the element at a specific rank.
# We'll find the (n//2)-th element for simplicity.

arr_copy_median = list(random_arr)
k_median = arr_size // 2
median_element = quick_select(arr_copy_median, 0, arr_size - 1, k_median)
# For verification, sort the original array and get the element at k_median-1 index
sorted_arr = sorted(random_arr)
verified_median = sorted_arr[k_median - 1]
print(f"Median element (k=n//2): {median_element} (Verified: {verified_median})")
```
