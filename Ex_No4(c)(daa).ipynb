```python
import heapq

def dijkstra(graph, source):
    """
    Dijkstra's Algorithm using Min-Heap, prioritizing fewer hops for equal total weights.
    Time: O((V + E) log V), Space: O(V)
    graph: dict {u: [(v, weight), ...]}, 0-indexed
    """
    n = len(graph)
    # dist stores (total_distance, hop_count)
    dist = [(float('inf'), float('inf'))] * n
    prev = [None] * n
    dist[source] = (0, 0)
    # pq stores (total_distance, hop_count, vertex)
    pq = [(0, 0, source)]
    visited = set()

    while pq:
        d_u, h_u, u = heapq.heappop(pq)

        # If we have already found a shorter path to u (or a path with same distance but fewer hops),
        # then skip this entry.
        if (d_u, h_u) > dist[u]:
            continue

        # Mark as visited only after processing its neighbors with the optimal path found so far
        # This handles cases where we might re-add nodes to PQ with better paths.

        for v, w in graph[u]:
            new_d = d_u + w
            new_h = h_u + 1

            current_d_v, current_h_v = dist[v]

            # Relaxation condition: prefer smaller distance, then smaller hop count
            if new_d < current_d_v or (new_d == current_d_v and new_h < current_h_v):
                dist[v] = (new_d, new_h)
                prev[v] = u
                heapq.heappush(pq, (new_d, new_h, v))

    return dist, prev

def reconstruct_path(prev, source, target):
    path = []
    node = target
    while node is not None:
        path.append(node)
        node = prev[node]
    path.reverse()
    if path and path[0] == source: # Ensure path starts from source
        return path
    return []

# --- Graph Definition (Adjacency List) ---
graph = {
    0: [(1, 4), (2, 1)],
    1: [(3, 1)],
    2: [(1, 2), (3, 5)],
    3: [(4, 3)],
    4: [(5, 2)],
    5: []
}

source = 0
dist, prev = dijkstra(graph, source)

print(f'Shortest paths from vertex {source}:')
print(f'{"Vertex":>8} {"Distance":>10} {"Hops":>5} {"Path":>30}')
print('-' * 60)
for v in range(len(graph)):
    path = reconstruct_path(prev, source, v)
    path_str = ' -> '.join(map(str, path)) if path else 'No path'

    d, h = dist[v]
    display_d = str(d) if d != float('inf') else 'INF'
    display_h = str(h) if h != float('inf') else 'INF'

    print(f'{v:>8} {display_d:>10} {display_h:>5} {path_str:>30}')
```
