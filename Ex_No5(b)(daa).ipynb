```python
import random

comparison_count = 0  # Global counter for original min_max_dc
comparison_count_extended = 0 # Global counter for extended min_max_dc

def min_max_dc(arr, low, high):
    global comparison_count
    # Base case: single element
    if low == high:
        return arr[low], arr[low]
    # Base case: two elements
    if high == low + 1:
        comparison_count += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        return arr[high], arr[low]

    # Divide
    mid = (low + high) // 2
    lmin, lmax = min_max_dc(arr, low, mid)
    rmin, rmax = min_max_dc(arr, mid + 1, high)

    # Conquer: combine with 2 comparisons
    comparison_count += 1
    overall_min = lmin if lmin < rmin else rmin
    comparison_count += 1
    overall_max = lmax if lmax > rmax else rmax
    return overall_min, overall_max

def min_max_dc_extended(arr, low, high):
    global comparison_count_extended

    # Base case: single element
    if low == high:
        # For a single element, all four are the same
        return arr[low], arr[low], arr[low], arr[low]

    # Base case: two elements
    if high == low + 1:
        # For two elements: arr[low], arr[high]
        comparison_count_extended += 1
        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low] # min1, min2, max1, max2
        else:
            return arr[high], arr[low], arr[low], arr[high] # min1, min2, max1, max2

    # Divide
    mid = (low + high) // 2
    lmin1, lmin2, lmax1, lmax2 = min_max_dc_extended(arr, low, mid)
    rmin1, rmin2, rmax1, rmax2 = min_max_dc_extended(arr, mid + 1, high)

    # Merge step to find overall min1, min2, max1, max2 (6 comparisons)

    # --- Find overall_min1 and overall_min2 ---
    overall_min1 = None
    overall_min2 = None

    # Find overall_min1 (1 comparison)
    comparison_count_extended += 1
    if lmin1 < rmin1:
        overall_min1 = lmin1
        # Candidates for overall_min2 are lmin2, rmin1, rmin2
        overall_min2 = lmin2 # Start with lmin2
        comparison_count_extended += 1
        if rmin1 < overall_min2: # Compare with rmin1
            overall_min2 = rmin1
        comparison_count_extended += 1
        if rmin2 < overall_min2: # Compare with rmin2
            overall_min2 = rmin2
    else: # rmin1 <= lmin1
        overall_min1 = rmin1
        # Candidates for overall_min2 are rmin2, lmin1, lmin2
        overall_min2 = rmin2 # Start with rmin2
        comparison_count_extended += 1
        if lmin1 < overall_min2: # Compare with lmin1
            overall_min2 = lmin1
        comparison_count_extended += 1
        if lmin2 < overall_min2: # Compare with lmin2
            overall_min2 = lmin2

    # --- Find overall_max1 and overall_max2 ---
    overall_max1 = None
    overall_max2 = None

    # Find overall_max1 (1 comparison)
    comparison_count_extended += 1
    if lmax1 > rmax1:
        overall_max1 = lmax1
        # Candidates for overall_max2 are lmax2, rmax1, rmax2
        overall_max2 = lmax2 # Start with lmax2
        comparison_count_extended += 1
        if rmax1 > overall_max2: # Compare with rmax1
            overall_max2 = rmax1
        comparison_count_extended += 1
        if rmax2 > overall_max2: # Compare with rmax2
            overall_max2 = rmax2
    else: # rmax1 >= lmax1
        overall_max1 = rmax1
        # Candidates for overall_max2 are rmax2, lmax1, lmax2
        overall_max2 = rmax2 # Start with rmax2
        comparison_count_extended += 1
        if lmax1 > overall_max2: # Compare with lmax1
            overall_max2 = lmax1
        comparison_count_extended += 1
        if lmax2 > overall_max2: # Compare with lmax2
            overall_max2 = lmax2

    return overall_min1, overall_min2, overall_max1, overall_max2

def min_max_naive(arr):
    mn, mx = arr[0], arr[0]
    comps = 0
    for x in arr[1:]:
        comps += 1
        if x < mn: mn = x
        comps += 1
        if x > mx: mx = x
    return mn, mx, comps

# --- Demonstration on small array ---
arr = [3, 1, 7, 4, 9, 2, 8, 5, 6, 0]

comparison_count = 0
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
dc_comps = comparison_count

comparison_count_extended = 0
mn1_ext, mn2_ext, mx1_ext, mx2_ext = min_max_dc_extended(arr, 0, len(arr) - 1)
dc_ext_comps = comparison_count_extended

_, _, naive_comps = min_max_naive(arr)

print(f'Array: {arr}')
print(f'Original D&C (Min-Max): Min: {mn}, Max: {mx}, Comparisons: {dc_comps}')
print(f'Extended D&C (Min1-Min2-Max1-Max2): Min1: {mn1_ext}, Min2: {mn2_ext}, Max1: {mx1_ext}, Max2: {mx2_ext}, Comparisons: {dc_ext_comps}')
print(f'Naive (Min-Max): Comparisons: {naive_comps}')

# --- Performance Analysis ---
# The formula for original D&C (Min-Max) is approximately (3/2)N - 2 for N being a power of 2.
# The formula for extended D&C (Min1-Min2-Max1-Max2) is approximately (7/2)N - 6 for N being a power of 2.
# This means roughly 2N - 4 additional comparisons for the extended version.
print(f'\n{"Size":>8} {"DC Comp":>10} {"Ext D&C Comp":>15} {"Naive Comp":>13} {"Formula 3n/2-2":>17} {"Formula 7n/2-6":>17}')
print('-' * 90)
for size in [10, 100, 1000, 10000]:
    arr = [random.randint(1, 10000) for _ in range(size)]

    comparison_count = 0
    mn, mx = min_max_dc(arr, 0, len(arr) - 1)
    dc_c = comparison_count

    comparison_count_extended = 0
    mn1_ext, mn2_ext, mx1_ext, mx2_ext = min_max_dc_extended(arr, 0, len(arr) - 1)
    dc_ext_c = comparison_count_extended

    _, _, naive_c = min_max_naive(arr)

    formula_orig = 3 * size // 2 - 2
    formula_ext = 7 * size // 2 - 6 # Approximation for N being a power of 2
    print(f'{size:>8} {dc_c:>10} {dc_ext_c:>15} {naive_c:>13} {formula_orig:>17} {formula_ext:>17}')

# Ex. No. 5 |                  To Find Min-Max Value by Applying Divide and Conquer Technique             CS5303 – DAA Lab
# Chennai Institute of Technology | Dept. of CSE | Page 20
```
