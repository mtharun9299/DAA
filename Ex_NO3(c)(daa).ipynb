### Prim's Algorithm with Adjacency Matrix (O(V^2))

```python
def prim_matrix(n, graph_matrix, start=0):
    """graph_matrix: adjacency matrix where graph_matrix[u][v] is weight of edge (u,v), INF if no edge"""
    INF    = float('inf')
    key    = [INF] * n
    parent = [-1]  * n
    inMST  = [False] * n

    key[start] = 0

    mst_edges = []
    mst_cost = 0

    for _ in range(n):
        # Find vertex with minimum key value, from the set of vertices not yet included in MST
        min_key = INF
        min_index = -1

        for v in range(n):
            if not inMST[v] and key[v] < min_key:
                min_key = key[v]
                min_index = v

        if min_index == -1: # Graph is not connected
            break

        u = min_index
        inMST[u] = True

        if parent[u] != -1:
            mst_edges.append((parent[u], u, graph_matrix[parent[u]][u]))
            mst_cost += graph_matrix[parent[u]][u]

        # Update key values and parent index of the adjacent vertices of the picked vertex.
        for v in range(n):
            # graph_matrix[u][v] is non zero only for adjacent vertices of u
            # inMST[v] is false for vertices not yet included in MST
            # Update the key only if graph_matrix[u][v] is smaller than key[v]
            if graph_matrix[u][v] < INF and not inMST[v] and graph_matrix[u][v] < key[v]:
                key[v] = graph_matrix[u][v]
                parent[v] = u

    return mst_edges, mst_cost
```

### Comparing Time Complexity for Dense Graphs (100 Vertices)

```python
import random
import time

# Generate a dense graph with 100 vertices
n_dense = 100
INF = float('inf')

# Adjacency Matrix for prim_matrix
graph_matrix_dense = [[INF] * n_dense for _ in range(n_dense)]
for i in range(n_dense):
    graph_matrix_dense[i][i] = 0 # No self-loops with weight

# Adjacency List for heap-based prim
adj_list_dense = {i: [] for i in range(n_dense)}

# Fill with random edges for a dense graph
# For a dense graph, connect almost every pair of vertices

# Generate edges once to ensure both algorithms use the same graph structure
edges_for_dense_graph = []
for i in range(n_dense):
    for j in range(i + 1, n_dense):
        if random.random() < 0.8: # 80% chance of an edge, making it dense
            weight = random.randint(1, 100)
            edges_for_dense_graph.append((weight, i, j))

# Populate adjacency matrix and list from generated edges
for w, u, v in edges_for_dense_graph:
    graph_matrix_dense[u][v] = w
    graph_matrix_dense[v][u] = w
    adj_list_dense[u].append((v, w))
    adj_list_dense[v].append((u, w))


print(f"Running Prim's algorithms on a dense graph with {n_dense} vertices...")

# Time the adjacency matrix version
start_time_matrix = time.time()
mst_matrix, cost_matrix = prim_matrix(n_dense, graph_matrix_dense)
end_time_matrix = time.time()

print(f"\nPrim's (Adjacency Matrix) Cost: {cost_matrix}, Time: {end_time_matrix - start_time_matrix:.6f} seconds")

# Time the heap-based version
start_time_heap = time.time()
mst_heap, cost_heap = prim(n_dense, adj_list_dense)
end_time_heap = time.time()

print(f"Prim's (Heap-based) Cost: {cost_heap}, Time: {end_time_heap - start_time_heap:.6f} seconds")

# Basic check that costs are the same (they should be for a connected graph)
if cost_matrix == cost_heap:
    print("\nBoth algorithms found MSTs with the same total cost.")
else:
    print("\nWarning: MST costs differ! Check graph connectivity or algorithm implementation.")
```
